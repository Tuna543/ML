```python
import seaborn as sns
import pandas as pd # Import pandas to ensure it's available for df
sns.set_theme(style="whitegrid")

# The df DataFrame already contains the data from '/content/Garment_Job_Risk.csv'
# Using sns.load_dataset() for a local file will raise an error.
# Instead, we will directly use the 'df' DataFrame that was loaded earlier.

# Draw a nested barplot by species and sex
g = sns.catplot(
    data=df, # Changed from 'penguins' to 'df'
    kind="bar",
    x="Garment_Name", y="Job_Role", hue="Working_Hours_Per_Week",
    errorbar="sd", palette="dark", alpha=.9, height=10
)
g.despine(left=True)
g.set_axis_labels("Garment Name", "Job Role") # Updated axis labels for relevance
g.legend.set_title("Working Hours Per Week") # Set a more descriptive legend title
```
