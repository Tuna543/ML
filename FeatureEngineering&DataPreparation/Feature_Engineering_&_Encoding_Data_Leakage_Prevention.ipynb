```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy = 'median')
df['App_Rating'] = imputer.fit_transform(df[['App_Rating']])
```

```python
df.isnull().sum()
```

```python
df['Match_Date'] = pd.to_datetime(df['Match_Date'])
```

```python
df['Match_DayOfWeek'] = df['Match_Date'].dt.day_name()
df = df.drop(columns= ['Match_Date'])
```

```python
df.head()
```

```python
df['Buffer_Per_Minute'] = df['Buffer_Count'] / (df['Watch_Minutes']+1)
```

```python
df.head()
```

```python
from sklearn.preprocessing import OrdinalEncoder
ord_enc = OrdinalEncoder(categories = [['Daily_Pass', 'Weekly_Pass', 'WC_Premium']])
df['Subscription_Plan'] = ord_enc.fit_transform(df[['Subscription_Plan']])
```

```python
df = pd.get_dummies(df, columns = ['Platform', 'Match_DayOfWeek', 'Login_Status'], drop_first= True, dtype =int)
```

```python
import category_encoders as ce

target_enc = ce.TargetEncoder(cols =['Phone_Prefix'])
df['Phone_Prefix'] = target_enc.fit_transform(df['Phone_Prefix'], df['Target_Uninstall'])
```

```python
from sklearn.preprocessing import StandardScaler, RobustScaler

scaler = RobustScaler()
df['Watch_Minutes'] = scaler.fit_transform(df[['Watch_Minutes']])
df['Buffer_Count'] = scaler.fit_transform(df[['Buffer_Count']])
df['Buffer_Per_Minute'] = scaler.fit_transform(df[['Buffer_Per_Minute']])
df['App_Rating'] = scaler.fit_transform(df[['App_Rating']])

```

```python
df.head()
```

##**Rule 1 of ML: Never evaluate your model on the data it trained on.**
##**Rule 2 of ML: Never let information from your test set "leak" into your training set.**

```python
#feature
X = df.drop(columns = ['Target_Uninstall'])

#label
y = df['Target_Uninstall']
```

```python
y.value_counts()
```

```python
!pip install imbalanced.learn
```

```python
from sklearn.model_selection import train_test_split
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state= 42)
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42, stratify=y)
```

```python
X_train_smote, y_train_smote = smote.fit_resample(X_train,y_train)
```

```python
from imblearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.metrics import classification_report
```

```python
pipeline = Pipeline(
    [
        ('scaler', RobustScaler()),
        ('smote', SMOTE(random_state=42)),
        ('model', RandomForestClassifier(random_state=42))
    ]
)

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

scores = cross_val_score(pipeline, X_train, y_train, cv=cv, scoring='accuracy')

# print("Cross-validation scores:", {scores} )
```

```python
pipeline.fit(X_train, y_train)

y_pred = pipeline.predict(X_test)
print(classification_report(y_test, y_pred))
```
