```python
# The Smart Battery Charger
# A device is charging. We monitor the percentage.
# To protect the battery, we slow down charging once it hits 80% and stop at 100%.

battery_level = 70
print(f'Starting charge at {battery_level}%')

while battery_level < 100:
    if battery_level < 80:
        battery_level += 5 # Fast charge
        mode = 'Fast Mode'
    else:
        battery_level += 2 # Trickle charge
        mode = 'Trickle Mode'

    # Cap at 100
    if battery_level > 100: battery_level = 100
    print(f'Status: {battery_level}% | Charging Mode: {mode}')

print('Battery fully charged. System idle.')
```

```python
# Advanced Story Problem: The Library Inventory
# A librarian is managing a queue of returned books.
# Some books are 'Damaged' and must be moved to a repair list.
# Others are 'New' and go to the front. The rest are shelved normally.

returns = [('Book A', 'Normal'), ('Book B', 'Damaged'), ('Book C', 'New'), ('Book D', 'Normal')]
shelve_queue = []
repair_list = []

for title, status in returns:
    if status == 'Damaged':
        repair_list.append(title)
    elif status == 'New':
        shelve_queue.insert(0, title)
    else:
        shelve_queue.append(title)

print(f'Books to Shelve (Priority First): {shelve_queue}')
print(f'Books for Repair: {repair_list}')
```

```python
#  The GPS Route Tracker
# A drone captures a series of GPS waypoints (Latitude, Longitude, Altitude).
# We need to calculate the total change in altitude across the flight path.

waypoints = [
    (40.7128, -74.0060, 100),
    (40.7130, -74.0065, 150),
    (40.7135, -74.0070, 120),
    (40.7140, -74.0075, 200)
]

total_ascent = 0
for i in range(len(waypoints) - 1):
    current_alt = waypoints[i][2]
    next_alt = waypoints[i+1][2]
    if next_alt > current_alt:
        total_ascent += (next_alt - current_alt)

print(f'Flight summary: {len(waypoints)} waypoints recorded.')
print(f'Total altitude gained: {total_ascent} meters.')
```

## 5. Sets

Formula: {item1, item2, ...}

Data: Unordered and Unique collection; holds only immutable (hashable) items; no duplicates allowed.

```python
# Simple Example: Basic Set Operations
numbers = {1, 2, 3, 3, 2, 1}
print(f'Unique numbers: {numbers}')
```

```python
#  Set Math (Union and Intersection)
math = {'Alice', 'Bob', 'Charlie'}
science = {'Bob', 'Dave', 'Eve'}

common = math.intersection(science)
print(f'Students in both math and science: {common}')

all = math.union(science)
print(f'All students: {all}')
```

## 7. Functions & Arguments

Formula: def (name)(parameters) → return (result)


Data: Accepts and returns Any Object (integers, strings, lists, or even other functions).

```python
# Simple Example: Basic Function
def greet(name):
    return f'Hello, {name}!'

print(greet('Sadman'))
```

```python
#  Default Arguments and Multiple Returns
def power_operations(base, exponent=2):
    square = base ** exponent
    cube = base ** 3
    return square, cube

sq, cb = power_operations(5)
print(f'Square: {sq}, Cube: {cb}')
```

## 8. List Comprehensions
Formula: [expression for item in iterable if condition]

ata: Produces a New List by processing elements from an existing iterable

```python
# Simple Example: Basic Squares
squares = [x**2 for x in range(1, 6)]
print(f'Squares: {squares}')
```

```python
# Medium Example: Conditional Filtering
words = ['apple', 'ice', 'banana', 'ink', 'cherry']
starts_with_vowel = [w for w in words if w[0] in 'aeiou']
print(f'Vowel words: {starts_with_vowel}')
```

```python
# Advanced Story Problem: The Sensor Data Cleaner
# An engineering team receives data from multiple sensors as a nested list.
# They need to flatten the list and remove any 'outlier' readings below 10 or above 100.

sensor_readings = [[12, 45, 105], [8, 99, 23], [50, 110, 42]]

# Flattening and filtering in one line
clean_data = [val for sublist in sensor_readings for val in sublist if 10 <= val <= 100]

print(f'Raw Data: {sensor_readings}')
print(f'Cleaned & Flattened Data: {clean_data}')
```

## 7. When and Why to Use Each Structure

* **Lists**: Ordered and mutable. Use for collections of items where order matters.
* **Tuples**: Ordered and immutable. Use for fixed data like coordinates.
* **Sets**: Unordered and unique. Use for membership testing and removing duplicates.
* **Dictionaries**: Key-value pairs. Use for fast lookups and structured data.
