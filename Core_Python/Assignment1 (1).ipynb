# Problem 1 (Student Grade Manager)
You are building a simple student grade management system. The program will collect student names and their scores, perform calculations, categorize their grades, and display a formatted report.

(a) Ask the user to enter their name and the number of subjects (between 1 and 5). Store these in appropriately named variables and print a welcome message using an f-	string.


```python
name=str(input("Enter your name: "))
total_subjects = int(input("Enter the number of total subjects: "))

print(f"Welcome to our website {name}.")
```

(b) Using a loop, collect a score (0–100) for each subject from the user and store all scores in a list. Validate each score. If the user enters a value outside 0–100, print an error message and ask again .


```python
user_all_marks = []

i=0
while True:
  if i==total_subjects:
    break

  marks = int(input('Enter the marks for your subject: '))

  if(marks<0 or marks>100):
    print(f"{marks} is not a valid marks. Please enter a valid marks.")
    continue

  user_all_marks.append(marks)
  i += 1

print(user_all_marks)
```

(c) Write a function calculate_stats(scores) that accepts the list as an argument and returns a tuple containing: total marks, average score, highest score, and lowest score.


```python
def calculate_stats(scores):
  i = 0
  info_tuple = ()
  total_marks = 0;
  highest_score = max(scores);
  lowest_score = min(scores);

  while i<len(scores):
    total_marks += scores[i]
    i += 1

  average_score = total_marks / len(scores)
  info_tuple = (total_marks, average_score, highest_score, lowest_score)
  return info_tuple

#calling the function to see result
total_marks, average_score, highest_score, lowest_score =calculate_stats(user_all_marks)
print(f'Total Marks: {total_marks}\nAverage score: {average_score}\nHighest score: {highest_score}\nLowest score: {lowest_score}\n')
```

(d) Write another function assign_grade(average), return a letter grade: A (≥80), B (≥65), C (≥50), D (≥35), F (below 35).


```python
def assign_grade(average):

  letter_grade = ''

  if average>79:
    letter_grade='A'
  elif average>64:
    letter_grade='B'
  elif average>49:
    letter_grade='C'
  elif average>34:
    letter_grade='D'
  else:
    letter_grade="F"

  return letter_grade

grade = assign_grade(average_score)
print(f'Gained grade is: {grade}')
```

(f) Print a neatly formatted report showing the student's name, all scores, total, average (rounded to 2 decimal places), highest, lowest, grade, and number of subjects passed.


```python
print('======================Grade Report=================================')
print(f'Student Name: {name}')
print(f'All scores in the {len(user_all_marks)} subjects: {user_all_marks}')
print(f'Total Score: {total_marks}')
print(f'Average Marks: {average_score:.2f}')
print(f'Highest Marks: {highest_score}')
print(f'Lowest Marks: {lowest_score}')
print(f'Letter Grade: {grade}')
print(f'Number of passed subject: {number_of_passed_subjects} out of {len(user_all_marks)}')
```

# Problem 2 (Inventory & Shopping Cart System)
You are building a mini inventory and shopping cart system for a small store. The system stores products and their details, allows a user to shop, and generates a summary at the end.


a) Create a dictionary called inventory with at least 6 products. Each key is the product name (string), and each value is another dictionary containing "price" (float) and "category" (string, e.g. "food", "electronics", "clothing").

```python
# inventory_info =
#  {'price': 100, 'category': 'food',
#   'price': 80, 'category' : 'food',
#   'price': 50000, 'category' : 'electronics',
#   'price': 60000, 'category' : 'electronics',
#   'price': 4500, 'category': 'clothing',
#   'price': 6560, 'category' : 'clothing'}
# inventory = {'Rice': inventory_info[100],
#              'Fruit' : inventory_info[80],
#              'Fridge' : inventory_info[50000],
#              'Television' : inventory_info[60000],
#              'Kurti' : inventory_info[4500],
#              'Agha Noor' : inventory_info[6560]}

inventory = {'Rice':
             {'price': 100, 'category': 'Food'},
             'Fruit':
             {'price': 80, 'category': 'Food'},
             'Fridge':
             {'price': 50000, 'category': 'Electronics'},
             'Television':
             {'price': 60000, 'category': 'Electronics'},
             'Kurti':
             {'price': 4500, 'category': 'Clothing'},
             'Agha Noor':
             {'price': 6500, 'category': 'Clothing'}
            }
```

(b) Extract all unique categories from the inventory. Print all available categories.

```python
unique_category = set()
for item, info in inventory.items():
  # print(f'Item: {item}\n')
  # print(f'Info: {info}\n')
  # print(f'Category: {info['category']}')
  unique_category.add(info['category'])

print(f'All unique items: {unique_category}')
```

(c) Display all available products and their prices. Then, using a loop, allow the user to keep adding product names to a list called cart until they type done. If they enter a product not in the inventory, print an appropriate error message and continue.

```python
#Displaying all available products

print('Our available products are shown below:')
for item, info in inventory.items():
  print(f'{item}, price is: {info['price']}')


unique_items = set()
for item, info in inventory.items():
  unique_items.add(item.lower())


cart = []
while(True):
  product_name = str(input("Enter the product name: "))
  product_name = product_name.lower()
  if product_name=="done":
    break
  if product_name in unique_items:
    cart.append(product_name.capitalize())
  else:
    print(f'Sorry there is not such product called {product_name} in our shop.')

print(f'Thanks for choosing us. You have added {len(cart)} products.')
if(len(cart)>0):
  print(f'Your added products are: {cart}.')
```

d) Write a function calculate_total(cart, inventory) that takes the cart list and inventory dictionary, computes the total price, and returns it. Call this function and display the result.


```python
def calculate_total(cart, inventory):
  total_price = 0
  for product in cart:
    # print(f'Product in cart: {product}')
    # print(f'Product in cart: {inventory.keys()}')
    if product in inventory.keys():
      total_price += inventory[product]['price'];
  return total_price;

total_price = calculate_total(cart, inventory)
print(f'Total price is: {total_price}')
```

(e) Apply a discount: if total ≥ 5000, apply 15% off; if total ≥ 2000, apply 10% off; if total ≥ 1000, apply 5% off; otherwise no discount. Print the discount applied and the final amount.


```python
def discount(total_price):
  discount_applied = 0;
  if total_price>=5000:
    discount_applied = total_price*.15
    total_price -= discount_applied
  elif total_price>=2000:
    discount_applied = total_price*.10
    total_price -= discount_applied
  elif total_price>=1000:
    discount_applied = total_price*.05
    total_price -= discount_applied

  return discount_applied, total_price;

discount_amount, price_after_discount = discount(total_price)
print(f'The discount amount is: {discount_amount}\nPrice_after_discount: {price_after_discount}')
```

(f) create a summary dict called cart_summary where keys are the unique items in the cart and values are how many times each item appears. Print this summary.


```python
def creating_summary_cart(cart):
  cart_summary = {}
  for i in range(len(cart)):
    if cart[i] in cart_summary.keys():
      cart_summary[cart[i]] += 1
    else:
      cart_summary[cart[i]] = 1
  return cart_summary

cart_summary = creating_summary_cart(cart)
print(f'Here is the summary cart: Items : How many times it appeared\n{cart_summary}')
```

g) Using a tuple, store the top 3 most expensive items from the entire inventory (by price) and print them. Use sorting — you are allowed to use sorted().

```python
def finding_three_expensive_items(inventory):
  copy_inventory = []
  for item, info in inventory.items():
    copy_inventory.append((item , inventory[item]['price']))
  copy_inventory = sorted(copy_inventory, key=lambda item: item[1])
  # copy_inventory.sort(reverse=True)
  return copy_inventory

three_expensive_items = finding_three_expensive_items(inventory)
print(f'Three expensive items: {three_expensive_items[-3:]}\n')
```

```python
#  The GPS Route Tracker
# A drone captures a series of GPS waypoints (Latitude, Longitude, Altitude).
# We need to calculate the total change in altitude across the flight path.

waypoints = [
    (40.7128, -74.0060, 100),
    (40.7130, -74.0065, 150),
    (40.7135, -74.0070, 120),
    (40.7140, -74.0075, 200)
]

total_ascent = 0
for i in range(len(waypoints) - 1):
    current_alt = waypoints[i][2]
    next_alt = waypoints[i+1][2]
    print(f'Current alt: {current_alt}\nNext alt: {next_alt}')
    if next_alt > current_alt:
        total_ascent += (next_alt - current_alt)

print(f'Flight summary: {len(waypoints)} waypoints recorded.')
print(f'Total altitude gained: {total_ascent} meters.')
```
